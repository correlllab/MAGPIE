```python
import numpy as np
import matplotlib.pyplot as plt
import json

# filepath = "../datasets/trajectories/1740511370.742381_white_paper_id-0/grasp.json"
# filepath = "../datasets/trajectories/1740511251.8399916_rubber_duck_id-0/grasp.json"
filepath_1 = "/home/will/MAGPIE/datasets/trajectories/1740693135.5613089_orange_id-0/grasp.json"
filepath_2 = "/home/will/MAGPIE/datasets/trajectories/1740693275.1866043_orange_id-0/grasp.json"

def plot_grasp(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)

    # plot spring constant over time
    # data is a list of dictionaries, each dictionary contains a spring constant at a certain timestamp

    forces = []
    time = []
    spring_constant = []
    apertures = []
    for d in data:
        time.append(d['timestamp'])
        forces.append(d['contact_force'])
        apertures.append(d['aperture'])
        spring_constant.append(d['k'])

    plt.plot(time, forces, label='contact force')
    plt.plot(time, apertures, label='aperture')
    print(apertures)
    # plt.plot(time, spring_constant, label='spring constant')
    # plot legend
    plt.legend()
    plt.show()

plot_grasp(filepath_1)
plot_grasp(filepath_2)
```
